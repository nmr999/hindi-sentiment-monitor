```
# Try MuRIL first; fall back to XLM-R if the download fails
import subprocess, sys
r = subprocess.run([sys.executable, 'src/finetune.py', '--source', 'data/hindi.csv', '--model', 'google/muril-base-cased', '--epochs', '4'])
if r.returncode != 0:
    print('MuRIL failed, falling back to xlm-roberta-base')
    subprocess.run([sys.executable, 'src/finetune.py', '--source', 'data/hindi.csv', '--model', 'xlm-roberta-base', '--epochs', '4'], check=True)
```
